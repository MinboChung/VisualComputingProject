## <center>Marching Tetrahedra</center>

The **Marching Tetrahedra** algorithm is an alternative method for isosurphase extraction in **volumetric data**.

It is designed to address topological ambiguities that can occur in Marching cubes by decomposing each cube into tetrahedra, which are simpler geometric units. 

### <center>1. Why use?</center>

- **Fixes ambiguities**: Marching cubes suffers from ambiguous cases where multiple triangulations are possible.
- **Better handling of surfaces**: Since tetrahedra are simple, they avoid incorrect surface connections in ambiguous configurations
- **More accurate representation**: Especially useful when preserving small details in the isosurface.

### <center>2. Algorithm overview</center>

#### Step 1: Decomposing the cube into tetrahedra
- Each cube in a 3D grid consists of 8 vertices.
- We divide this cube into 5 or 6 tetrahedra (Different schemes exist).
- Each tetrahedron is treated separately for isosurface intersection calcultions.

#### Step 2: Evaluating the Isosurface Condition
- Each tetrahedron has 4 vertices, each with a scalar value
- We determine which of these vertices are above and below the isosurface threshold
- If some vertices are above and others below, the isosurface must pass through the tetrahedron

#### Step 3: Finding Intersection Points
- The isosurface cuts through tetrahedron edges where scalar values change from below to above the threshold
- We interpolate points along these edges to determine where the surface crosses.

#### Step 4: Generating triangles
- Each tetrahedron can form 0, 1, or 2 triangles, depending on how the isosurface intersects the tetrahedron.
- The resulting triangles approximate the **isosurface**.

### <center>3. Cube decomposition schemes</center>

#### 2 common methods to decompose a cube into tetrahedra
1. 5-Tetrahedra Decomposition
   1. The cube is split into 5s, each sharing a central vertex
   2. Simpler method but introduces slight asymmetry
2. 6-Tetrahedra Decomposition (Preferred)
   1. The cube is divided into 6 tetrahedra by splitting along the cube's diagronals
   2. This method is more balanced and widely used in practice.
